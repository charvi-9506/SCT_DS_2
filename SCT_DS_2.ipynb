```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
sns.set_theme(style="whitegrid")
survival_palette = {0: "#34495e", 1: "#2ecc71"}
```

```python
df = pd.read_csv('train.csv')

print("="*60)
print("TITANIC DATASET - CLEANING & EDA")
print("="*60)
```

```python
print(f"\n1. DATASET OVERVIEW\nShape: {df.shape}\n\nMissing Values Before:\n{df.isnull().sum()}")
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df.drop(columns=['Cabin'], errors='ignore', inplace=True)
```

```python
df['FamilySize'] = df[['SibSp', 'Parch']].sum(axis=1) + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
df['Title'] = df['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)
```
